```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Résultats des modèles
results = {
    "MLP": {
        "Ra": {"R2": 0.6188, "MAE": 1.6008, "RMSE": 2.3000, "MAPE": 21.06},
        "Density": {"R2": 0.5372, "MAE": 1.6163, "RMSE": 2.8199, "MAPE": 1.74},
    },
    "XGBoost": {
        "Ra": {"R2": 0.6945, "MAE": 1.4155, "RMSE": 2.0590, "MAPE": 18.32},
        "Density": {"R2": 0.5550, "MAE": 1.8221, "RMSE": 2.7650, "MAPE": 1.94},
    },
    "CatBoost": {
        "Ra": {"R2": 0.7286, "MAE": 1.2048, "RMSE": 1.9409, "MAPE": 15.68},
        "Density": {"R2": 0.4916, "MAE": 1.6334, "RMSE": 2.9553, "MAPE": 1.77},
    },
    "TabNet": {
        "Ra": {"R2": 0.6114, "MAE": 1.4884, "RMSE": 2.3224, "MAPE": 19.51},
        "Density": {"R2": 0.5374, "MAE": 1.5455, "RMSE": 2.8192, "MAPE": 1.68},
    },
    "Random Forest": {
        "Ra": {"R2": 0.1934, "MAE": 1.8793, "RMSE": 3.3457, "MAPE": 27.51},
        "Density": {"R2": 0.4517, "MAE": 1.7797, "RMSE": 3.0691, "MAPE": 1.93},
    }
}

# Conversion en DataFrame
metrics = []
for model, targets in results.items():
    for target, values in targets.items():
        for metric, value in values.items():
            metrics.append({
                'Model': model,
                'Target': target,
                'Metric': metric,
                'Value': value
            })

metrics_df = pd.DataFrame(metrics)

# Création des graphiques en barres
sns.set(style="whitegrid")
for metric_name in ["R2", "MAE", "RMSE", "MAPE"]:
    plt.figure(figsize=(10, 6))
    plot_df = metrics_df[metrics_df['Metric'] == metric_name]
    sns.barplot(data=plot_df, x="Model", y="Value", hue="Target")
    plt.title(f"Comparaison des modèles - {metric_name}")
    plt.ylabel(metric_name)
    plt.xlabel("Modèle")
    plt.legend(title="Cible")
    plt.tight_layout()
    plt.show()
```
